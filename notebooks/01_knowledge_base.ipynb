```python
from pathlib import Path
import pandas as pd
PROJECT_ROOT = Path.cwd().parent
KNOWLEDGE_BASE_DIR = PROJECT_ROOT / "data" / "knowledge_base"

print("Project root:", PROJECT_ROOT)
print("Knowledge base:", KNOWLEDGE_BASE_DIR)
```

```python
files = sorted(KNOWLEDGE_BASE_DIR.glob("*.md"))

print(f"Number of documents: {len(files)}")

for file in files:
    print("-", file.name)
```

```python
for file in files:
    print("=" * 70)
    print(f"FILE: {file.name}")
    print("=" * 70)

    text = file.read_text(encoding="utf-8")

    print(text[:1000])
    print()
```

```python
documents = []

for file in files:
    text = file.read_text(encoding="utf-8").strip()

    documents.append({
        "source": file.name,
        "text": text
    })

print(f"Loaded documents: {len(documents)}")
print()

for document in documents:
    print(document["source"], "→", len(document["text"]), "characters")
```

```python
documents[0]
```

```python
lengths = [len(document["text"]) for document in documents]

print("Minimum characters:", min(lengths))
print("Maximum characters:", max(lengths))
print("Average characters:", sum(lengths) / len(lengths))
```

```python
document_summary = pd.DataFrame({
    "source": [document["source"] for document in documents],
    "characters": lengths
})

document_summary
```

```python
document_summary = document_summary.sort_values(
    "characters",
    ascending=False
).reset_index(drop=True)

document_summary
```

```python
print("Total documents:", len(documents))
print("Total characters:", sum(lengths))
print("Average document length:", round(sum(lengths) / len(lengths), 2))
```
