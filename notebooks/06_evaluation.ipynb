```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

retrieval_df = pd.read_csv(
    PROCESSED_DIR / "retrieval_test_results.csv"
)

print("Retrieval test cases:", len(retrieval_df))
retrieval_df
```

```python
expected_sources = [
    "account.md",
    "returns.md",
    "refunds.md",
    "technical_support.md",
    "billing.md"
]

retrieval_df["expected_source"] = expected_sources

retrieval_df["correct"] = (
    retrieval_df["source"] == retrieval_df["expected_source"]
)

retrieval_df
```

```python
retrieval_accuracy = retrieval_df["correct"].mean()

print(f"Retrieval Accuracy: {retrieval_accuracy:.2%}")
```

```python
evaluation_summary = {
    "total_questions": len(retrieval_df),
    "correct_retrievals": int(retrieval_df["correct"].sum()),
    "retrieval_accuracy": retrieval_accuracy
}

evaluation_summary
```

```python
import json

evaluation_path = PROCESSED_DIR / "evaluation_summary.json"

with open(evaluation_path, "w", encoding="utf-8") as f:
    json.dump(evaluation_summary, f, indent=4)

print("Saved:", evaluation_path)
```

```python
print("Final Retrieval Accuracy:", f"{retrieval_accuracy:.2%}")
```
